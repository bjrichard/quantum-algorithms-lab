# 03_measurements_and_povms

Purpose:
- Contrast projective measurements with Positive Operator-Valued Measures (POVMs)
- Understand measurement as an operational process
- See why POVMs are natural in NISQ devices

```python
# Projective measurement in the Z basis using density matrices (no simulators).
# p(0) = Tr(rho |0><0|),  p(1) = Tr(rho |1><1|)

import numpy as np
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector, DensityMatrix

# 1) Prepare a state (example: |+> via Hadamard on |0>)
qc = QuantumCircuit(1)
qc.h(0)

psi = Statevector.from_instruction(qc)
rho = DensityMatrix(psi).data  # use the explicit matrix for trace calculations

# 2) Define Z-basis projectors
P0 = np.array([[1, 0],
               [0, 0]], dtype=complex)  # |0><0|
P1 = np.array([[0, 0],
               [0, 1]], dtype=complex)  # |1><1|

# 3) Compute probabilities via trace formula
p0 = np.real(np.trace(rho @ P0))
p1 = np.real(np.trace(rho @ P1))

print("rho:\n", rho)
print("\nZ-basis measurement probabilities:")
print("p(0) =", p0)
print("p(1) =", p1)
print("p(0)+p(1) =", p0 + p1)
```

```python
# Single-qubit POVM example:
# E± = (I ± η Z) / 2
# Compute outcome probabilities p(±) = Tr(rho E±)

import numpy as np
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector, DensityMatrix

# --- 1) Prepare a state (example: |+>) ---
qc = QuantumCircuit(1)
qc.h(0)

psi = Statevector.from_instruction(qc)
rho = DensityMatrix(psi).data  # explicit matrix form

# --- 2) Define operators ---
I = np.eye(2, dtype=complex)
Z = np.array([[1, 0],
              [0, -1]], dtype=complex)

# --- 3) POVM sharpness parameter ---
eta = 0.6   # values between 0 and 1

E_plus  = 0.5 * (I + eta * Z)
E_minus = 0.5 * (I - eta * Z)

# --- 4) Sanity checks ---
print("E+ + E- =\n", E_plus + E_minus)
print("Eigenvalues of E+:", np.linalg.eigvals(E_plus))
print("Eigenvalues of E-:", np.linalg.eigvals(E_minus))

# --- 5) Compute outcome probabilities ---
p_plus  = np.real(np.trace(rho @ E_plus))
p_minus = np.real(np.trace(rho @ E_minus))

print("\nPOVM outcome probabilities:")
print("p(+) =", p_plus)
print("p(-) =", p_minus)
print("p(+)+p(-) =", p_plus + p_minus)
```

Key takeaway:
- Projective measurements are a special case of POVMs
- POVMs describe measurement statistics without assuming ideal collapse
- NISQ readout noise is naturally captured by POVMs
