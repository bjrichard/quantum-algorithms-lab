# 02_density_matrices

Purpose:
- Understand density matrices as statistical descriptions of quantum states
- Contrast pure states vs mixed states
- Connect physical intuition to computational representations

```python
# Construct a single-qubit pure state using a Hadamard gate
# and compute its density matrix.

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector, DensityMatrix

# 1) Create a single-qubit circuit
qc = QuantumCircuit(1)

# 2) Apply a Hadamard gate: |0> -> |+>
qc.h(0)

# 3) Obtain the statevector corresponding to the circuit
psi = Statevector.from_instruction(qc)

# 4) Construct the density matrix rho = |psi><psi|
rho = DensityMatrix(psi)

# 5) Print results clearly
print("Statevector |psi>:\n")
print(psi.data)

print("\nDensity matrix rho = |psi><psi|:\n")
print(rho.data)
```

```python
import numpy as np
from qiskit.quantum_info import DensityMatrix, Statevector, Operator

# --- Pure state density matrix (|+><+|) from the previous step ---
psi_plus = Statevector([1/np.sqrt(2), 1/np.sqrt(2)])
rho_plus = DensityMatrix(psi_plus)

# --- Mixed state: 50/50 classical mixture of |0> and |1> ---
rho_0 = DensityMatrix(Statevector([1, 0]))  # |0><0|
rho_1 = DensityMatrix(Statevector([0, 1]))  # |1><1|
rho_mix = 0.5 * rho_0 + 0.5 * rho_1

print("rho_plus (|+><+|):\n", rho_plus.data)
print("\n rho_mix (0.5|0><0| + 0.5|1><1|):\n", rho_mix.data)

# Quick sanity checks
print("\nTrace rho_plus:", np.trace(rho_plus.data))
print("Trace rho_mix :", np.trace(rho_mix.data))
```

```python
from qiskit.quantum_info import SparsePauliOp

X = SparsePauliOp.from_list([("X", 1.0)])
Z = SparsePauliOp.from_list([("Z", 1.0)])

def expval(rho, op):
    # <op> = Tr(rho * op)
    return np.real(np.trace(rho.data @ op.to_matrix()))

print("<Z> for rho_plus:", expval(rho_plus, Z))
print("<Z> for rho_mix :", expval(rho_mix, Z))

print("<X> for rho_plus:", expval(rho_plus, X))
print("<X> for rho_mix :", expval(rho_mix, X))
```

```python
import numpy as np

def purity(rho):
    return np.real(np.trace(rho.data @ rho.data))

print("Purity (pure |+>):", purity(rho_plus))
print("Purity (mixed 50/50):", purity(rho_mix))
```
